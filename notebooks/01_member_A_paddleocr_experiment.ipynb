# Tuần 1 - Thử nghiệm Nhận dạng Biển số xe với PaddleOCR
**Đề tài**: Hệ thống Quản lý Xe Ra Vào Đại học Vinh Bằng Thị giác Máy tính  
**Thành viên A**: AI & Computer Vision Core

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt
import re
from paddleocr import PaddleOCR

# Khởi tạo PaddleOCR Engine
ocr = PaddleOCR(use_angle_cls=True, lang='en')
```

## 1. Kiểm tra trên Biển số Ô tô (1 dòng)

```python
car_path = '../data/test_samples/car_sample.jpg'
img = cv2.imread(car_path)
plt.figure(figsize=(6, 2))
plt.imshow(cv2.cvtColor(img, cv2.COLOR_BGR2RGB))
plt.axis('off')
plt.show()

results = ocr.ocr(car_path, cls=True)
for line in results[0]:
    print(f"Nhận dạng: {line[1][0]} (Độ tin cậy: {line[1][1]:.2%})")
```

## 2. Kiểm tra trên Biển số Xe máy (2 dòng)

```python
moto_path = '../data/test_samples/moto_sample.jpg'
img_moto = cv2.imread(moto_path)
plt.figure(figsize=(4, 3))
plt.imshow(cv2.cvtColor(img_moto, cv2.COLOR_BGR2RGB))
plt.axis('off')
plt.show()

results_moto = ocr.ocr(moto_path, cls=True)
for line in results_moto[0]:
    print(f"Dòng đọc được: {line[1][0]} (Độ tin cậy: {line[1][1]:.2%})")
```
